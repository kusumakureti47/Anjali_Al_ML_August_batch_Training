#missing value analysis



```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
total_missing = df.isnull().sum().sum()

print("Total missing values:", total_missing)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing value percentage:")
print(missing_percentage)
```

```python
most_missing_column = df.isnull().sum().idxmax()
most_missing_count = df.isnull().sum().max()

print("Column with the highest missing values:", most_missing_column)
print("Number of missing values:", most_missing_count)
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numerical columns:")
print(numeric_columns.tolist())
```

```python
column = "lifetime_spend_groceries"

Q1 = df[column].quantile(0.25)
Q3 = df[column].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

print("Sample customer names:")
print(df["customer_name"].head(20))

```python
outliers = df[
    (df[column] < lower_bound) |
    (df[column] > upper_bound)
]

print("Number of potential outliers:", len(outliers))
```

```python
values = df["lifetime_spend_groceries"].dropna().to_numpy()

print("Mean:", np.mean(values))
print("Median:", np.median(values))
print("Minimum:", np.min(values))
print("Maximum:", np.max(values))
print("Standard deviation:", np.std(values))
```

```python
print("Rows before cleaning:", len(original_df))
print("Rows after cleaning:", len(df))

print("Missing values before cleaning:",
      original_df.isnull().sum().sum())

print("Missing values after cleaning:",
      df.isnull().sum().sum())

print("Duplicates before cleaning:",
      original_df.duplicated().sum())

print("Duplicates after cleaning:",
      df.duplicated().sum())
```
